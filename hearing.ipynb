```python
# Imports and helpers
import re
import json
from pathlib import Path
from docx import Document
from docx.text.paragraph import Paragraph
from docx.table import Table
from IPython.display import JSON, display

# Regex patterns
HEADER_RE = re.compile(r"^(ක්‍රියාකාරකම\s*[-–]?\s*\d+)\s*$", re.IGNORECASE)
NUMBER_RE = re.compile(r"^(\d{1,2})\b(?:[.)\-]?\s*)(.*)$")
SINHALA_TOKEN_RE = re.compile(r"[\u0D80-\u0DFF]+(?:[\-–—]?[\u0D80-\u0DFF]+)*")

def iter_doc_blocks(doc: Document):
    """
    Yield text blocks (strings) in the order they appear in the document body,
    covering both paragraphs and table cells.
    Each yielded block is stripped.
    """
    body = doc.element.body
    for child in body.iterchildren():
        # paragraph
        if child.tag.endswith('}p'):
            para = Paragraph(child, doc)
            text = para.text.strip()
            if text:
                yield text
        # table
        elif child.tag.endswith('}tbl'):
            tbl = Table(child, doc)
            for r in tbl.rows:
                for c in r.cells:
                    # split lines inside a cell and yield each non-empty line as a block
                    for line in c.text.splitlines():
                        line = line.strip()
                        if line:
                            yield line

def extract_tokens_from_text(text):
    """Return list of Sinhala tokens from text (preserve order)."""
    return [m.group(0) for m in SINHALA_TOKEN_RE.finditer(text)]
```

```python
# Parsing function (core logic)
def parse_blocks(blocks):
    """
    Given a list of ordered text blocks (paragraphs and table cells),
    return parsed structure mapping headers -> { items: [...], tokens: [...] }.
    """
    result = {}
    current_header = None
    i = 0
    n = len(blocks)
    while i < n:
        block = blocks[i].strip()
        # detect header
        h = HEADER_RE.match(block)
        if h:
            # normalize header to remove internal whitespace so keys look like "ක්‍රියාකාරකම-01"
            header_key = re.sub(r"\s+", "", h.group(1))
            current_header = header_key
            if current_header not in result:
                result[current_header] = {"items": [], "tokens": []}
            i += 1
            continue

        # if inside a section, look for numbered entry
        if current_header:
            m = NUMBER_RE.match(block)
            if m:
                idx = m.group(1).zfill(2)
                remainder = m.group(2).strip()
                words_line = remainder

                # if number line has no words, look ahead for next 1-2 blocks as the word(s)
                if not words_line:
                    collected = []
                    j = i + 1
                    while j < n and len(collected) < 2:
                        nxt = blocks[j].strip()
                        # if next block starts a new header or number, stop
                        if HEADER_RE.match(nxt) or NUMBER_RE.match(nxt):
                            break
                        if nxt:
                            collected.append(nxt)
                        j += 1
                    words_line = " ".join(collected).strip()
                    i = j  # advance past the blocks we consumed
                else:
                    i += 1

                # extract Sinhala tokens (preserve exact strings)
                tokens = extract_tokens_from_text(words_line)
                if len(tokens) >= 2:
                    words_pair = tokens[:2]
                elif len(tokens) == 1:
                    words_pair = [tokens[0]]
                else:
                    # fallback: maybe the block itself contains tokens
                    fb = extract_tokens_from_text(block)
                    words_pair = fb[:2] if fb else []

                result[current_header]["items"].append({"index": idx, "words": words_pair})
                for t in words_pair:
                    result[current_header]["tokens"].append(t)
                continue

        # not a header/number: move on
        i += 1

    return result
```

```python
# Main execution cell: load docx, parse, save JSON, and display a preview
DOCX_PATH = Path("docs\swara.docx")  # change path if needed
OUT_JSON = Path("swara_from_docx_notebook.json")

if not DOCX_PATH.exists():
    raise FileNotFoundError(f"{DOCX_PATH} not found. Upload swara.docx to the notebook folder or change DOCX_PATH.")

doc = Document(DOCX_PATH)
blocks = [b for b in iter_doc_blocks(doc) if b.strip() != ""]

print(f"Found {len(blocks)} text blocks (paragraph lines & table cells).")
# Optional: show first 30 blocks for debugging
# for k,bl in enumerate(blocks[:30],1): print(f"{k:02d} | {bl}")

parsed = parse_blocks(blocks)

if not parsed:
    print("No sections parsed. Showing first 50 blocks for debugging:")
    for k, b in enumerate(blocks[:50], start=1):
        print(f"{k:02d} | {b!r}")
else:
    # Save JSON
    OUT_JSON.write_text(json.dumps(parsed, ensure_ascii=False, indent=2), encoding="utf-8")
    print("Saved JSON to:", OUT_JSON.resolve())
    # Display a readable preview in the notebook (first section)
    first_key = next(iter(parsed))
    print("\nPreview (first section):", first_key)
    display(JSON(parsed[first_key]))
```

```python
# Improved Sinhala -> Singlish transliteration (notebook cell)
import json
from pathlib import Path
import unicodedata

IN_JSON = Path("swara_from_docx_notebook.json")
OUT_JSON = Path("swara_translit_improved.json")

if not IN_JSON.exists():
    raise FileNotFoundError(f"Input JSON not found: {IN_JSON}")

# --- mapping: consonant -> base latin consonant (no trailing vowel) ---
CONS = {
    "ක":"k","ඛ":"kh","ග":"g","ඝ":"gh","ඟ":"ng",
    "ච":"ch","ඡ":"chh","ජ":"j","ඣ":"jh","ඤ":"ny",
    "ට":"t","ඨ":"th","ඩ":"d","ඪ":"dh","ණ":"n",
    "ත":"t","ථ":"th","ද":"d","ධ":"dh","න":"n",
    "ප":"p","ඵ":"ph","බ":"b","භ":"bh","ම":"m",
    "ය":"y","ර":"r","ල":"l","ව":"v","ශ":"sh",
    "ෂ":"sh","ස":"s","හ":"h","ළ":"l","ෆ":"f",
    # extras
    "ඥ":"gn","ඦ":"gn"
}

# independent vowels
INDEP_V = {
    "අ":"a","ආ":"aa","ඇ":"ae","ඈ":"aae",
    "ඉ":"i","ඊ":"ii","උ":"u","ඌ":"uu",
    "එ":"e","ඒ":"ee","ඔ":"o","ඕ":"oo",
    "ඓ":"ai","ඖ":"au"
}

# vowel signs (matras) that follow consonants
V_SIGN = {
    "ා":"aa",  # long a
    "ි":"i",   # i
    "ී":"ii",  # long i
    "ු":"u",   # u
    "ූ":"uu",  # long u
    "ෙ":"e",   # e
    "ේ":"ee",  # long e
    "ෛ":"ai",
    "ො":"o",
    "ෝ":"oo",
    "ෞ":"au",
    "ෘ":"ru",  # approximation
    "ෲ":"ruu"
}

DIAC = {
    "ං":"n",   # anusvara
    "ඃ":"h"
}

VIRAMA = "්"

# sets for fast checks
CONS_SET = set(CONS.keys())
INDEP_SET = set(INDEP_V.keys())
V_SIGN_SET = set(V_SIGN.keys())
DIAC_SET = set(DIAC.keys())

def normalize_text(s: str) -> str:
    # NFC normal form
    return unicodedata.normalize("NFC", s)

def transliterate_word(word: str, cap_first: bool = True) -> str:
    w = normalize_text(word)
    i = 0
    n = len(w)
    out = []

    while i < n:
        ch = w[i]

        # independent vowel
        if ch in INDEP_SET:
            out.append(INDEP_V[ch])
            i += 1
            continue

        # consonant + possible sequences
        if ch in CONS_SET:
            # gemination: C + virama + same C  -> double consonant
            # e.g., ල + ෍ (virama) + ල  (i.e. ල්ලා)
            if i+2 < n and w[i+1] == VIRAMA and w[i+2] == ch:
                base = CONS[ch]  # base consonant translit
                # consume C, virama, second C
                i += 3
                # after the doubled consonant may come a vowel sign or diacritic
                vowel_part = ""
                while i < n and (w[i] in V_SIGN_SET or w[i] in DIAC_SET):
                    if w[i] in V_SIGN_SET:
                        vowel_part = V_SIGN[w[i]]
                    elif w[i] in DIAC_SET:
                        vowel_part += DIAC[w[i]]
                    i += 1
                # if no explicit vowel sign -> implicit 'a'
                if vowel_part == "":
                    vowel_part = "a"
                out.append(base + base + vowel_part)  # doubled consonant + vowel
                continue

            # normal consonant handling
            base = CONS[ch]
            i += 1
            vowel_part = ""
            had_vowel_sign = False
            # if next is virama -> pure consonant (no vowel)
            if i < n and w[i] == VIRAMA:
                # consume virama, produce base (no vowel)
                i += 1
                out.append(base)  # no trailing vowel
                continue

            # collect vowel signs/diacritics immediately following consonant
            while i < n and (w[i] in V_SIGN_SET or w[i] in DIAC_SET):
                if w[i] in V_SIGN_SET:
                    vowel_part = V_SIGN[w[i]]
                    had_vowel_sign = True
                elif w[i] in DIAC_SET:
                    vowel_part += DIAC[w[i]]
                i += 1

            # if no vowel sign, implicit 'a'
            if vowel_part == "":
                vowel_part = "a"
            out.append(base + vowel_part)
            continue

        # vowel sign or diacritic alone (rare) - map if possible
        if ch in V_SIGN_SET:
            out.append(V_SIGN[ch])
            i += 1
            continue
        if ch in DIAC_SET:
            out.append(DIAC[ch])
            i += 1
            continue

        # anything else (punctuation, ASCII) passes through
        out.append(ch)
        i += 1

    result = "".join(out)
    if cap_first and result:
        result = result[0].upper() + result[1:]
    return result

# quick test set (compare expected)
test_words = [
    "අම්මා", "පස්", "පුස්", "හත", "හිත", "බල්ලා", "බිල්ලා",
    "බය", "බිය", "පනා", "පිනා", "ටින්", "ටෝන්", "පිල්", "ටපාල්"
]
print("Test transliterations:")
for tw in test_words:
    print(tw, "->", transliterate_word(tw))

# load input JSON, transliterate and save
data = json.loads(IN_JSON.read_text(encoding="utf-8"))
outd = {}
for sec, payload in data.items():
    outd[sec] = {"items": [], "tokens": []}
    for it in payload.get("items", []):
        idx = it.get("index", "")
        words = it.get("words", [])
        twords = [transliterate_word(w, cap_first=True) for w in words]
        outd[sec]["items"].append({"index": idx, "words": twords})
        outd[sec]["tokens"].extend(twords)

OUT_JSON.write_text(json.dumps(outd, ensure_ascii=False, indent=2), encoding="utf-8")
print("\nWrote:", OUT_JSON.resolve())
```

```python
# Notebook cell: extract vowel of the FIRST Sinhala letter/syllable for each word and add Singlish transliteration
import json, csv, unicodedata
from pathlib import Path
from typing import Tuple
 
IN_JSON = Path("swara_from_docx_notebook.json")
OUT_CSV = Path("swara_words_vowels_firstletter.csv")
OUT_JSON = Path("swara_words_vowels_firstletter.json")
 
# independent vowels
INDEP_VOWELS = {
    "අ":"a","ආ":"aa","ඇ":"ae","ඈ":"aae",
    "ඉ":"i","ඊ":"ii","උ":"u","ඌ":"uu",
    "එ":"e","ඒ":"ee","ඔ":"o","ඕ":"oo",
    "ඓ":"ai","ඖ":"au"
}
# dependent vowel signs -> (independent char, short name)
VOWEL_SIGN_TO_INDEP = {
    "ා": ("ආ","aa"), "ි": ("ඉ","i"), "ී": ("ඊ","ii"),
    "ු": ("උ","u"), "ූ": ("ඌ","uu"), "ෙ": ("එ","e"),
    "ේ": ("ඒ","ee"), "ෛ": ("ඓ","ai"), "ො": ("ඔ","o"),
    "ෝ": ("ඕ","oo"), "ෞ": ("ඖ","au"), "ෘ": ("උ (approx)","ru"),
    "ෲ": ("ඌ (approx)","ruu")
}
 
# --- Singlish transliteration mappings ---
CONS = {
    "ක":"k","ඛ":"kh","ග":"g","ඝ":"gh","ඟ":"ng",
    "ච":"ch","ඡ":"chh","ජ":"j","ඣ":"jh","ඤ":"ny",
    "ට":"t","ඨ":"th","ඩ":"d","ඪ":"dh","ණ":"n",
    "ත":"t","ථ":"th","ද":"d","ධ":"dh","න":"n",
    "ප":"p","ඵ":"ph","බ":"b","භ":"bh","ම":"m",
    "ය":"y","ර":"r","ල":"l","ව":"v","ශ":"sh",
    "ෂ":"sh","ස":"s","හ":"h","ළ":"l","ෆ":"f",
    "ඥ":"gn","ඦ":"gn"  # extras
}
 
V_SIGN = {
    "ා":"aa",  # long a
    "ි":"i",   # i
    "ී":"ii",  # long i
    "ු":"u",   # u
    "ූ":"uu",  # long u
    "ෙ":"e",   # e
    "ේ":"ee",  # long e
    "ෛ":"ai",
    "ො":"o",
    "ෝ":"oo",
    "ෞ":"au",
    "ෘ":"ru",  # approximation
    "ෲ":"ruu"
}
 
DIAC = {
    "ං":"n",   # anusvara
    "ඃ":"h"
}
 
VIRAMA = "්"
 
# sets for fast checks
INDEP_SET = set(INDEP_VOWELS.keys())
SIGN_SET = set(VOWEL_SIGN_TO_INDEP.keys())
CONS_SET = set(CONS.keys())
V_SIGN_SET = set(V_SIGN.keys())
DIAC_SET = set(DIAC.keys())
 
def normalize_text(s: str) -> str:
    return unicodedata.normalize("NFC", s)
 
def first_letter_vowel(word: str) -> Tuple[str,str]:
    """
    Return (vowel_char, vowel_name) for the FIRST Sinhala letter/syllable of `word`.
    Rules:
      - If first char is independent vowel -> return it.
      - If first char is consonant:
         * if immediately followed by a dependent vowel sign -> return that mapped vowel
         * if immediately followed by virama or nothing -> return implicit 'අ' (a)
      - If first char not Sinhala -> return ("","")
    """
    w = normalize_text(word).strip()
    if not w:
        return "",""
 
    # find first Sinhala character that's either vowel or consonant
    first_char = w[0]
 
    # if first char itself is an independent vowel
    if first_char in INDEP_SET:
        return first_char, INDEP_VOWELS[first_char]
 
    # if first char is a Sinhala consonant (approx by Unicode block)
    # we'll check combining signs that follow; treat any non-matching as implicit 'a'
    # A simple way: check next character if it's a vowel sign or virama.
    # We assume the word is NFC-normalized so vowel signs are separate codepoints after base consonant.
    if '\u0D80' <= first_char <= '\u0DFF':
        # check the immediate next codepoint(s)
        if len(w) >= 2:
            nxt = w[1]
            # vowel sign present immediately after first consonant
            if nxt in SIGN_SET:
                indep_char, name = VOWEL_SIGN_TO_INDEP[nxt]
                return indep_char, name
            # virama present -> explicit consonant with no vowel; user requested "always vowel"
            if nxt == VIRAMA:
                return "අ", "a"
            # if next char is another consonant or diacritic or ascii, default implicit 'අ'
            return "අ", "a"
        else:
            # single-char consonant -> implicit 'අ'
            return "අ", "a"
 
    # first char not Sinhala
    return "",""
 
def transliterate_word(word: str, cap_first: bool = True) -> str:
    """Convert Sinhala word to Singlish transliteration."""
    w = normalize_text(word)
    i = 0
    n = len(w)
    out = []
 
    while i < n:
        ch = w[i]
 
        # independent vowel
        if ch in INDEP_SET:
            out.append(INDEP_VOWELS[ch])
            i += 1
            continue
 
        # consonant + possible sequences
        if ch in CONS_SET:
            # gemination: C + virama + same C  -> double consonant
            if i+2 < n and w[i+1] == VIRAMA and w[i+2] == ch:
                base = CONS[ch]
                i += 3
                vowel_part = ""
                while i < n and (w[i] in V_SIGN_SET or w[i] in DIAC_SET):
                    if w[i] in V_SIGN_SET:
                        vowel_part = V_SIGN[w[i]]
                    elif w[i] in DIAC_SET:
                        vowel_part += DIAC[w[i]]
                    i += 1
                if vowel_part == "":
                    vowel_part = "a"
                out.append(base + base + vowel_part)
                continue
 
            # normal consonant handling
            base = CONS[ch]
            i += 1
            vowel_part = ""
            if i < n and w[i] == VIRAMA:
                i += 1
                out.append(base)
                continue
 
            while i < n and (w[i] in V_SIGN_SET or w[i] in DIAC_SET):
                if w[i] in V_SIGN_SET:
                    vowel_part = V_SIGN[w[i]]
                elif w[i] in DIAC_SET:
                    vowel_part += DIAC[w[i]]
                i += 1
 
            if vowel_part == "":
                vowel_part = "a"
            out.append(base + vowel_part)
            continue
 
        # vowel sign or diacritic alone
        if ch in V_SIGN_SET:
            out.append(V_SIGN[ch])
            i += 1
            continue
        if ch in DIAC_SET:
            out.append(DIAC[ch])
            i += 1
            continue
 
        # anything else passes through
        out.append(ch)
        i += 1
 
    result = "".join(out)
    if cap_first and result:
        result = result[0].upper() + result[1:]
    return result
 
# --- build rows ---
if not IN_JSON.exists():
    raise FileNotFoundError(f"Input JSON not found: {IN_JSON}")
 
data = json.loads(IN_JSON.read_text(encoding="utf-8"))
rows = []
for section, payload in data.items():
    for it in payload.get("items", []):
        idx = it.get("index", "")
        words_field = it.get("words") if "words" in it else it.get("word")
        if words_field is None:
            continue
        words_list = [words_field] if isinstance(words_field, str) else words_field
        for w in words_list:
            if not w:
                vowel_char, vowel_name = "",""
                singlish = ""
            else:
                vowel_char, vowel_name = first_letter_vowel(w)
                singlish = transliterate_word(w)
            rows.append({
                "section": section,
                "index": idx,
                "sinhala_word": w,
                "singlish_word": singlish,
                "main_vowel_char": vowel_char,
                "main_vowel_name": vowel_name
            })
 
# --- write CSV and JSON ---
with OUT_CSV.open("w", encoding="utf-8", newline="") as f:
    writer = csv.DictWriter(f, fieldnames=["section", "index", "sinhala_word", "singlish_word", "main_vowel_char", "main_vowel_name"])
    writer.writeheader()
    writer.writerows(rows)
 
OUT_JSON.write_text(json.dumps(rows, ensure_ascii=False, indent=2), encoding="utf-8")
print("Wrote:", OUT_CSV.resolve())
print("Wrote:", OUT_JSON.resolve())
 
# quick preview of output structure
for r in rows[:3]:
    print("\nSample row:", r)
```

```python
# SINGLE NOTEBOOK CELL: Generate high-quality Sinhala audio per-word, sectioned folders.
# - Uses Google Cloud TTS (WaveNet) if GOOGLE_APPLICATION_CREDENTIALS is set.
# - Falls back to gTTS + pydub if not.
# - Post-processes with pydub: slow down slightly + increase volume.
# - Allows pronunciation overrides for words that are unclear.
#
# Install dependencies first (run in a notebook cell if needed):
# !pip install google-cloud-texttospeech gTTS pydub tqdm unidecode

import os, json, io, re
from pathlib import Path
from tqdm import tqdm
from unidecode import unidecode
from pydub import AudioSegment

# Try to import Google Cloud TTS
USE_GOOGLE_CLOUD = False
try:
    from google.cloud import texttospeech
    # If GOOGLE_APPLICATION_CREDENTIALS env var present, enable cloud usage
    if os.environ.get("GOOGLE_APPLICATION_CREDENTIALS"):
        USE_GOOGLE_CLOUD = True
except Exception:
    USE_GOOGLE_CLOUD = False

# Fallback gTTS
from gtts import gTTS

# ---------- Config ----------
INPUT_JSON = Path("swara_words_vowels_firstletter.json")   # change if your JSON has a different name
OUTPUT_ROOT = Path("swara_audio_cloud")                    # top-level output folder
LANG = "si"  # for gTTS fallback
# Post-processing
SLOW_FACTOR = 1.15       # 1.15 => ~15% slower (tweak to taste)
VOLUME_GAIN_DB = 5       # increase volume by +5 dB
# Google Cloud TTS params (only used if USE_GOOGLE_CLOUD True)
GCP_VOICE_NAME = "si-LK-Wavenet-A"   # example; adjust if needed or list available voices
GCP_AUDIO_ENCODING = texttospeech.AudioEncoding.MP3 if USE_GOOGLE_CLOUD else None
GCP_SPEAKING_RATE = 0.95   # 0.9-1.0 = slightly slower; fine tune
GCP_PITCH = 0.0            # adjust if voice sounds too low/high

# Pronunciation overrides: map original Sinhala word -> alternate text to send to TTS
# Use this to manually fix words with unclear pronunciation.
# Example: "පිනා": "පි-නා" or a small transliteration/phonetic form; try different forms until pronunciation is good.
pronunciation_overrides = {
    # "පිනා": "පි-නා",
    # "කජු": "ක-ජු",
    # add problematic words here and experiment
}

# ---------- Helpers ----------
def safe_unicode_filename(s: str) -> str:
    s = s.strip()
    s = re.sub(r"[\\/:\*\?\"<>\|]", "_", s)
    s = re.sub(r"\s+", "_", s)
    return s[:60]

def ascii_translit(s: str) -> str:
    return re.sub(r"[^\w\-_.]", "", unidecode(s)) or "word"

def change_speed(sound: AudioSegment, factor: float) -> AudioSegment:
    """Return a copy of sound slowed by `factor` (factor >1 -> slower)."""
    # pydub change via frame rate trick
    new_frame_rate = int(sound.frame_rate / factor)
    slowed = sound._spawn(sound.raw_data, overrides={"frame_rate": new_frame_rate})
    return slowed.set_frame_rate(sound.frame_rate)

def save_audiosegment_to_mp3(audio: AudioSegment, path: Path, bitrate="192k"):
    audio.export(str(path), format="mp3", bitrate=bitrate)

# ---------- Load input JSON ----------
if not INPUT_JSON.exists():
    raise FileNotFoundError(f"Input JSON not found: {INPUT_JSON.resolve()}")

records = json.loads(INPUT_JSON.read_text(encoding="utf-8"))

# Build grouping by section -> list of items (preserve given order)
from collections import OrderedDict
sections = OrderedDict()
for r in records:
    sec = r.get("section", "unknown")
    sections.setdefault(sec, []).append(r)

# Make output structure
OUTPUT_ROOT.mkdir(parents=True, exist_ok=True)

# If using Google Cloud TTS client, initialize
gcp_client = None
if USE_GOOGLE_CLOUD:
    try:
        gcp_client = texttospeech.TextToSpeechClient()
    except Exception as e:
        print("Google Cloud TTS import/initialization failed:", e)
        USE_GOOGLE_CLOUD = False

# ---------- Generation loop ----------
total_items = sum(len(v) for v in sections.values())
pbar = tqdm(total=total_items, desc="Generating audio")

for sec, items in sections.items():
    sec_folder = OUTPUT_ROOT / sec
    sec_folder.mkdir(parents=True, exist_ok=True)

    for item in items:
        idx = str(item.get("index", "")).zfill(2)
        word = item.get("sinhala_word", "").strip()
        if not word:
            pbar.update(1)
            continue

        # Use override text if provided
        tts_text = pronunciation_overrides.get(word, word)

        uni_name = safe_unicode_filename(word)
        ascii_name = ascii_translit(word)
        out_filename = f"{idx}_{uni_name}_{ascii_name}.mp3"
        out_path = sec_folder / out_filename

        # Skip if already exists
        if out_path.exists():
            pbar.update(1)
            continue

        try:
            if USE_GOOGLE_CLOUD and gcp_client:
                # Prepare request: supports SSML if needed (we use plain text now)
                input_text = texttospeech.SynthesisInput(text=tts_text)
                # voice selection: pick Sinhala WaveNet voice if available; otherwise pick language code
                voice = texttospeech.VoiceSelectionParams(language_code="si-LK", name=GCP_VOICE_NAME)
                audio_config = texttospeech.AudioConfig(
                    audio_encoding=GCP_AUDIO_ENCODING,
                    speaking_rate=GCP_SPEAKING_RATE,
                    pitch=GCP_PITCH
                )
                response = gcp_client.synthesize_speech(input=input_text, voice=voice, audio_config=audio_config)
                # response.audio_content is raw bytes (MP3)
                audio_bytes = response.audio_content
                # load into pydub for post-processing
                audio = AudioSegment.from_file(io.BytesIO(audio_bytes), format="mp3")
            else:
                # gTTS fallback
                # Use slow=False to let us control speed via pydub; we slow down after generation.
                tts = gTTS(text=tts_text, lang=LANG, slow=False)
                buf = io.BytesIO()
                tts.write_to_fp(buf)
                buf.seek(0)
                audio = AudioSegment.from_file(buf, format="mp3")

            # Post-process: slow a bit and increase volume
            processed = change_speed(audio, SLOW_FACTOR)
            processed = processed + VOLUME_GAIN_DB

            # Save final mp3
            save_audiosegment_to_mp3(processed, out_path)

        except Exception as e:
            # On error, print and attempt a minimal fallback filename using ascii-only
            print(f"Error generating audio for '{word}' in {sec}: {e}")
            fallback_out = sec_folder / f"{idx}_{ascii_name}.mp3"
            try:
                # Try gTTS again as last resort
                tts = gTTS(text=tts_text, lang=LANG, slow=False)
                buf = io.BytesIO(); tts.write_to_fp(buf); buf.seek(0)
                audio = AudioSegment.from_file(buf, format="mp3")
                processed = change_speed(audio, SLOW_FACTOR)
                processed = processed + VOLUME_GAIN_DB
                save_audiosegment_to_mp3(processed, fallback_out)
            except Exception as e2:
                print("Fallback also failed for", word, ":", e2)

        pbar.update(1)

pbar.close()
print("Done. Audio saved under:", OUTPUT_ROOT.resolve())
print("NOTE: To improve pronunciation for specific words, edit `pronunciation_overrides` and re-run the cell.")
```

```python
# Simple transliteration (ascii fallback) for filename / display
from unidecode import unidecode
import re

def ascii_translit(s: str) -> str:
    return re.sub(r"[^\w\-_.]", "", unidecode(s)) or "word"

def safe_audio_path(section: str, idx: str, sinhala_word: str) -> str:
    # Build a relative path to audio file if exists (assumes you generated audio)
    uni_name = re.sub(r"[\\/:\*\?\"<>\|]", "_", sinhala_word).strip()
    filename = f"{idx}_{uni_name}_{ascii_translit(sinhala_word)}.mp3"
    return str(DEFAULT_AUDIO_ROOT / section / filename)
```

```python
# DB abstraction: either in-memory dict or mongomock
class TestDB:
    def __init__(self, use_mongomock=False):
        self.use_mongomock = use_mongomock
        if use_mongomock:
            client = mongomock.MongoClient()
            self.col = client.test.swara
        else:
            # simple dict store: key -> doc
            self._store = {}
            self.col = None

    def insert_many(self, docs: List[Dict[str, Any]]):
        if self.use_mongomock:
            self.col.insert_many(docs)
        else:
            for d in docs:
                self._store[d["_id"]] = d

    def find(self, query=None):
        if self.use_mongomock:
            return list(self.col.find(query or {}))
        else:
            if not query:
                return list(self._store.values())
            # naive query support for equality on top-level keys only
            res=[]
            for d in self._store.values():
                ok=True
                for k,v in (query or {}).items():
                    if d.get(k) != v:
                        ok=False; break
                if ok: res.append(d)
            return res

    def find_one(self, query):
        if self.use_mongomock:
            return self.col.find_one(query)
        else:
            for d in self._store.values():
                ok=True
                for k,v in query.items():
                    if d.get(k) != v:
                        ok=False; break
                if ok: return d
            return None

    def update_one(self, query, update: Dict[str, Any]):
        # update is dict of fields to set (like {"$set": {...}})
        if self.use_mongomock:
            self.col.update_one(query, update)
            return
        doc = self.find_one(query)
        if not doc:
            return
        if "$set" in update:
            for k,v in update["$set"].items():
                doc[k]=v
        if "$push" in update:
            for k,v in update["$push"].items():
                doc.setdefault(k,[]).append(v)

    def replace_one(self, query, new_doc):
        if self.use_mongomock:
            self.col.replace_one(query, new_doc)
            return
        doc = self.find_one(query)
        if not doc:
            return
        # find key and replace
        for _id,d in list(self._store.items()):
            if all(d.get(k)==v for k,v in query.items()):
                self._store[_id]=new_doc
                return
```

```python
# Helper: vowel ordering (you can adjust or compute unique vowels from data)
VOWEL_ORDER = ["a","i","u","e","oo","ii","uu","ai","au"]  # example order — adapt to your test plan

def get_words_by_vowel(vowel_name: str, section: Optional[str]=None):
    query = {}
    if section:
        query["section"]=section
    all_docs = db.find(query)
    return [d for d in all_docs if d.get("main_vowel_name")==vowel_name]

def get_default_word():
    # default: first item in ක්‍රියාකාරකම-01 index 01
    return db.find_one({"section":"ක්‍රියාකාරකම-01","index":"01"})
```

```python
# Core workflow functions used by the frontend/backend

def present_word_to_user(word_doc: Dict[str,Any]):
    """Simulated presentation: return the info frontend would show."""
    return {
        "id": word_doc["_id"],
        "section": word_doc["section"],
        "index": word_doc["index"],
        "sinhala": word_doc["sinhala_word"],
        "singlish": word_doc["singlish"],
        "audio": word_doc["audio_path"],
        "vowel": word_doc["main_vowel_name"]
    }

def record_user_event(word_id: str, event: Dict[str,Any]):
    """Push user event onto the doc's user_results and update vowel_tested flag."""
    db.update_one({"_id": word_id}, {"$push":{"user_results": event}})
    db.update_one({"_id": word_id}, {"$set":{"vowel_tested": True}})

def mark_vowel_status(vowel_name: str, status: str, section: Optional[str]=None):
    """Mark ALL words containing vowel_name in section with vowel_status (for simulation).
       In real app this would be per-user state; here we mark docs for quick inspection."""
    query = {}
    if section:
        query["section"]=section
    docs = db.find(query)
    for d in docs:
        if d.get("main_vowel_name")==vowel_name:
            db.update_one({"_id": d["_id"]}, {"$set": {"vowel_status": status}})

def submit_response(word_id: str, choice: str):
    """
    choice: 'heard_correct' | 'did_not_hear' | 'chosen_other' (if user chooses alternative)
    Implements double-confirmation:
       - If 'heard_correct': present another word with same vowel.
           - If second also 'heard_correct' -> vowel == 'heard' (success) and move to next vowel.
           - Else -> treat as not-heard (or require third check) — we use confirmation logic here.
       - If 'did_not_hear': present another word with same vowel to confirm inability:
           - If second 'did_not_hear' -> mark vowel 'not_heard' and move on.
    Returns: dict with next_action and payload describing what to show next.
    """
    doc = db.find_one({"_id": word_id})
    if not doc:
        return {"error": "word not found"}

    vowel = doc.get("main_vowel_name")
    section = doc.get("section")

    # record first attempt
    record_user_event(word_id, {"choice": choice})

    # check how many events exist for this vowel in this section (to decide confirmation)
    same_vowel_docs = get_words_by_vowel(vowel, section=section)

    # find how many events for any doc with this vowel in same section
    events_for_vowel = []
    for d in same_vowel_docs:
        for ev in d.get("user_results",[]):
            if ev.get("choice") in ("heard_correct","did_not_hear"):
                events_for_vowel.append((d["_id"], ev))

    # If user said heard_correct on first try -> need confirmation using another word with same vowel
    if choice == "heard_correct":
        # find a different word with same vowel that hasn't been used yet
        candidate = None
        for d in same_vowel_docs:
            if d["_id"] == word_id: continue
            if not d.get("user_results"):
                candidate = d; break
        if not candidate:
            # if no unused candidate, pick any other
            candidate = next((d for d in same_vowel_docs if d["_id"] != word_id), None)

        if candidate:
            # return instruction to frontend to present this confirm word
            return {"next_action":"present_confirm_word", "word": present_word_to_user(candidate)}
        else:
            # nothing to confirm with -> tentatively mark as heard
            mark_vowel_status(vowel, "heard", section=section)
            return {"next_action":"vowel_marked", "vowel": vowel, "status":"heard"}

    if choice == "did_not_hear":
        # similar confirm strategy but for inability
        candidate = None
        for d in same_vowel_docs:
            if d["_id"] == word_id: continue
            if not d.get("user_results"):
                candidate = d; break
        if not candidate:
            candidate = next((d for d in same_vowel_docs if d["_id"] != word_id), None)
        if candidate:
            return {"next_action":"present_confirm_word", "word": present_word_to_user(candidate)}
        else:
            mark_vowel_status(vowel, "not_heard", section=section)
            return {"next_action":"vowel_marked", "vowel": vowel, "status":"not_heard"}

    if choice == "confirm_heard":  # user confirmed second word was heard
        mark_vowel_status(vowel, "heard", section=section)
        return {"next_action":"vowel_marked","vowel":vowel,"status":"heard"}

    if choice == "confirm_not_heard":  # user confirmed second word was also not heard
        mark_vowel_status(vowel, "not_heard", section=section)
        return {"next_action":"vowel_marked","vowel":vowel,"status":"not_heard"}

    # other choices can be implemented (e.g. user selecting the correct among multiple options)
    return {"next_action":"noop"}
```

```python
# Simulate an interactive session programmatically.
# In real frontend you'd call get_default_word(), play audio, capture user choice, call submit_response(), etc.

def simulate_user_hears_first_then_confirms(default_word_doc, user_hears_first=True):
    # simulate frontend: present default
    print("Presenting default:", default_word_doc["sinhala_word"], "(", default_word_doc["_id"], ")")
    if user_hears_first:
        resp = submit_response(default_word_doc["_id"], "heard_correct")
        print("After first response:", resp)
        if resp["next_action"] == "present_confirm_word":
            candidate = resp["word"]
            # simulate user confirming again
            resp2 = submit_response(candidate["id"], "confirm_heard")
            print("After confirmation:", resp2)
    else:
        resp = submit_response(default_word_doc["_id"], "did_not_hear")
        print("After first not-heard:", resp)
        if resp["next_action"] == "present_confirm_word":
            candidate = resp["word"]
            resp2 = submit_response(candidate["id"], "confirm_not_heard")
            print("After confirmation not-heard:", resp2)

# pick default
default = get_default_word()
simulate_user_hears_first_then_confirms(default, user_hears_first=True)

# inspect DB state for vowel 'a' in section 01
print("\nDocs for vowel 'a' in ක්‍රියාකාරකම-01:")
pp.pprint(get_words_by_vowel("a", section="ක්‍රියාකාරකම-01")[:5])
```

```python
# Automated sweep: for each vowel in VOWEL_ORDER, try first available word and simulate user choices.
def auto_test_flow(user_behavior_fn):
    # user_behavior_fn(word_doc) -> returns True if user hears it, False if not
    results = {}
    for vowel in VOWEL_ORDER:
        # find a section+word for this vowel (search across all sections)
        words = get_words_by_vowel(vowel)
        if not words:
            continue
        # pick the first word for testing
        first_word = words[0]
        hears = user_behavior_fn(first_word)
        if hears:
            r = submit_response(first_word["_id"], "heard_correct")
            # if confirm needed
            if r.get("next_action") == "present_confirm_word":
                confirm_word = r["word"]
                # decide behavior for confirm (we can have different behavior or same)
                confirm_hears = user_behavior_fn(confirm_word)
                if confirm_hears:
                    submit_response(confirm_word["id"], "confirm_heard")
                    results[vowel] = "heard"
                else:
                    # user failed confirmation -> mark not heard
                    submit_response(confirm_word["id"], "confirm_not_heard")
                    results[vowel] = "not_heard"
            else:
                results[vowel] = r.get("status","unknown")
        else:
            r = submit_response(first_word["_id"], "did_not_hear")
            if r.get("next_action")=="present_confirm_word":
                confirm_word = r["word"]
                confirm_hears = user_behavior_fn(confirm_word)
                if not confirm_hears:
                    submit_response(confirm_word["id"], "confirm_not_heard")
                    results[vowel]="not_heard"
                else:
                    submit_response(confirm_word["id"], "confirm_heard")
                    results[vowel]="heard"
            else:
                results[vowel]=r.get("status","unknown")
    return results

# Example user behavior function: simulate user who hears vowels 'a' and 'i' only
def user_behavior_example(word_doc):
    # simple rule: user hears vowels 'a' and 'i' only
    return word_doc.get("main_vowel_name") in ("a","i")

res = auto_test_flow(user_behavior_example)
print("Auto test results per vowel:", res)

# Inspect some DB entries
print("\nSample docs after auto test:")
for d in db.find()[:8]:
    pp.pprint(d)
```

```python

# Notebook cell: extract vowel of the FIRST Sinhala letter/syllable for each word,
# add Singlish transliteration and produce portable relative audio_path
import json, csv, unicodedata, re, os
from pathlib import Path
from typing import Tuple
from unidecode import unidecode
 
# ---------- CONFIG ----------
IN_JSON = Path("swara_from_docx_notebook.json")
OUT_CSV = Path("swara_words_vowels_firstletter.csv")
OUT_JSON = Path("swara_words_vowels_firstletter.json")
 
# set this to the local audio folder inside your project.
# final stored audio_path will start with AUDIO_BASE_PATH.name (e.g. "swara_audio_cloud/...")
AUDIO_BASE_PATH = Path("hearing_project_v0.01/swara_audio_cloud")
 
# ---------- vowel and transliteration maps (kept from your code) ----------
INDEP_VOWELS = {
    "අ":"a","ආ":"aa","ඇ":"ae","ඈ":"aae",
    "ඉ":"i","ඊ":"ii","උ":"u","ඌ":"uu",
    "එ":"e","ඒ":"ee","ඔ":"o","ඕ":"oo",
    "ඓ":"ai","ඖ":"au"
}
VOWEL_SIGN_TO_INDEP = {
    "ා": ("ආ","aa"), "ි": ("ඉ","i"), "ී": ("ඊ","ii"),
    "ු": ("උ","u"), "ූ": ("ඌ","uu"), "ෙ": ("එ","e"),
    "ේ": ("ඒ","ee"), "ෛ": ("ඓ","ai"), "ො": ("ඔ","o"),
    "ෝ": ("ඕ","oo"), "ෞ": ("ඖ","au"), "ෘ": ("උ (approx)","ru"),
    "ෲ": ("ඌ (approx)","ruu")
}
CONS = {
    "ක":"k","ඛ":"kh","ග":"g","ඝ":"gh","ඟ":"ng",
    "ච":"ch","ඡ":"chh","ජ":"j","ඣ":"jh","ඤ":"ny",
    "ට":"t","ඨ":"th","ඩ":"d","ඪ":"dh","ණ":"n",
    "ත":"t","ථ":"th","ද":"d","ධ":"dh","න":"n",
    "ප":"p","ඵ":"ph","බ":"b","භ":"bh","ම":"m",
    "ය":"y","ර":"r","ල":"l","ව":"v","ශ":"sh",
    "ෂ":"sh","ස":"s","හ":"h","ළ":"l","ෆ":"f",
    "ඥ":"gn","ඦ":"gn"
}
V_SIGN = {
    "ා":"aa","ි":"i","ී":"ii","ු":"u","ූ":"uu",
    "ෙ":"e","ේ":"ee","ෛ":"ai","ො":"o","ෝ":"oo","ෞ":"au",
    "ෘ":"ru","ෲ":"ruu"
}
DIAC = {"ං":"n","ඃ":"h"}
VIRAMA = "්"
 
INDEP_SET = set(INDEP_VOWELS.keys())
SIGN_SET = set(VOWEL_SIGN_TO_INDEP.keys())
CONS_SET = set(CONS.keys())
V_SIGN_SET = set(V_SIGN.keys())
DIAC_SET = set(DIAC.keys())
 
def normalize_text(s: str) -> str:
    return unicodedata.normalize("NFC", s)
 
def first_letter_vowel(word: str) -> Tuple[str,str]:
    w = normalize_text(word).strip()
    if not w:
        return "",""
    first_char = w[0]
    if first_char in INDEP_SET:
        return first_char, INDEP_VOWELS[first_char]
    if '\u0D80' <= first_char <= '\u0DFF':
        if len(w) >= 2:
            nxt = w[1]
            if nxt in SIGN_SET:
                indep_char, name = VOWEL_SIGN_TO_INDEP[nxt]
                return indep_char, name
            if nxt == VIRAMA:
                return "අ", "a"
            return "අ", "a"
        else:
            return "අ", "a"
    return "",""
 
# ---------- filename helpers (produce expected filenames) ----------
def safe_unicode_filename(s: str) -> str:
    s = (s or "").strip()
    s = re.sub(r"[\\/:\*\?\"<>\|]", "_", s)
    s = re.sub(r"\s+", "_", s)
    return s[:60]
 
def ascii_translit(s: str) -> str:
    out = re.sub(r"[^\w\-_.]", "", unidecode(s) or "")
    return out or "word"
 
def expected_filenames(section: str, idx: str, sinhala_word: str):
    idx2 = str(idx).zfill(2)
    uni = safe_unicode_filename(sinhala_word)
    asc = ascii_translit(sinhala_word)
    # two common patterns
    return [f"{idx2}_{uni}_{asc}.mp3", f"{idx2}_{uni}.mp3"]
 
# ---------- get relative portable audio path ----------
def get_portable_audio_path(section: str, idx: str, sinhala_word: str) -> str:
    """
    Return a portable relative path starting from AUDIO_BASE_PATH.name.
    If file exists physically under AUDIO_BASE_PATH, return a true relative path
    (e.g. "swara_audio_cloud/ක්‍රියාකාරකම-01/01_පස්_ps.mp3").
    If not found, return constructed relative path (same format) so the app knows where to look.
    """
    section_folder = AUDIO_BASE_PATH / section
    for fname in expected_filenames(section, idx, sinhala_word):
        candidate = section_folder / fname
        if candidate.exists():
            try:
                # attempt to compute path relative to current working directory
                rel = os.path.relpath(candidate, start=Path.cwd())
                # normalize to forward slashes and strip any leading "./"
                rel = rel.replace("\\", "/").lstrip("./")
                # ensure it starts with AUDIO_BASE_PATH.name; if not, construct
                if not rel.startswith(AUDIO_BASE_PATH.name):
                    # produce constructed path beginning with AUDIO_BASE_PATH.name
                    return f"{AUDIO_BASE_PATH.name}/{section}/{fname}"
                return rel
            except Exception:
                # fallback constructed relative path
                return f"{AUDIO_BASE_PATH.name}/{section}/{fname}"
    # if no candidate exists, still return constructed path so DB knows expected location
    # (this helps other machines know where to place audio files)
    # choose first expected filename for the constructed path
    fname = expected_filenames(section, idx, sinhala_word)[0]
    return f"{AUDIO_BASE_PATH.name}/{section}/{fname}"
 
# ---------- transliteration (same as your function) ----------
def transliterate_word(word: str, cap_first: bool = True) -> str:
    w = normalize_text(word)
    i = 0
    n = len(w)
    out = []
    while i < n:
        ch = w[i]
        if ch in INDEP_SET:
            out.append(INDEP_VOWELS[ch]); i += 1; continue
        if ch in CONS_SET:
            # gemination (C + virama + same C)
            if i+2 < n and w[i+1] == VIRAMA and w[i+2] == ch:
                base = CONS[ch]; i += 3
                vowel_part = ""
                while i < n and (w[i] in V_SIGN_SET or w[i] in DIAC_SET):
                    if w[i] in V_SIGN_SET: vowel_part = V_SIGN[w[i]]
                    elif w[i] in DIAC_SET: vowel_part += DIAC[w[i]]
                    i += 1
                if vowel_part == "": vowel_part = "a"
                out.append(base + base + vowel_part); continue
            base = CONS[ch]; i += 1
            vowel_part = ""
            if i < n and w[i] == VIRAMA:
                i += 1; out.append(base); continue
            while i < n and (w[i] in V_SIGN_SET or w[i] in DIAC_SET):
                if w[i] in V_SIGN_SET: vowel_part = V_SIGN[w[i]]
                elif w[i] in DIAC_SET: vowel_part += DIAC[w[i]]
                i += 1
            if vowel_part == "": vowel_part = "a"
            out.append(base + vowel_part); continue
        if ch in V_SIGN_SET:
            out.append(V_SIGN[ch]); i += 1; continue
        if ch in DIAC_SET:
            out.append(DIAC[ch]); i += 1; continue
        out.append(ch); i += 1
    result = "".join(out)
    if cap_first and result:
        result = result[0].upper() + result[1:]
    return result
 
# ---------- build rows ----------
if not IN_JSON.exists():
    raise FileNotFoundError(f"Input JSON not found: {IN_JSON}")
 
data = json.loads(IN_JSON.read_text(encoding="utf-8"))
rows = []
for section, payload in data.items():
    for it in payload.get("items", []):
        idx = it.get("index", "")
        words_field = it.get("words") if "words" in it else it.get("word")
        if words_field is None:
            continue
        words_list = [words_field] if isinstance(words_field, str) else words_field
        for w in words_list:
            if not w:
                vowel_char, vowel_name = "",""
                singlish = ""
                audio_path = ""
            else:
                vowel_char, vowel_name = first_letter_vowel(w)
                singlish = transliterate_word(w)
                audio_path = get_portable_audio_path(section, idx, w)
            rows.append({
                "section": section,
                "index": idx,
                "sinhala_word": w,
                "singlish_word": singlish,
                "main_vowel_char": vowel_char,
                "main_vowel_name": vowel_name,
                "audio_path": audio_path
            })
 
# ---------- write CSV and JSON ----------
with OUT_CSV.open("w", encoding="utf-8", newline="") as f:
    writer = csv.DictWriter(f, fieldnames=["section", "index", "sinhala_word", "singlish_word", "main_vowel_char", "main_vowel_name", "audio_path"])
    writer.writeheader()
    writer.writerows(rows)
 
OUT_JSON.write_text(json.dumps(rows, ensure_ascii=False, indent=2), encoding="utf-8")
print("Wrote:", OUT_CSV.resolve())
print("Wrote:", OUT_JSON.resolve())
 
# quick preview of output structure
for r in rows[:6]:
    print("\nSample row:", r)
```
